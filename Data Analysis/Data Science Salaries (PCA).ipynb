```python
# Importation des bibliothèques nécessaires
import pandas as pd
from sklearn.preprocessing import MinMaxScaler
from sklearn.decomposition import PCA
from sklearn.manifold import TSNE

# 1) Chargement du dataset Data Science Job Salaries
url = "https://raw.githubusercontent.com/user257814938/PSTB-DI-Bootcamp/refs/heads/main/Week2/Day3/DailyChallenge/datascience_salaries.csv"
df = pd.read_csv(url)

# Vérification du contenu du dataset
print("Aperçu du dataset :")
print(df.head())

# 2) Normalisation de la colonne 'salary' avec Min-Max
scaler = MinMaxScaler()
df["salary_normalized"] = scaler.fit_transform(df[["salary"]])

print("\nAperçu des valeurs normalisées (0-1) :")
print(df[["salary", "salary_normalized"]].head())

# 3) Réduction de dimensions (PCA)
# Sélection des colonnes numériques pour PCA
numeric_cols = df.select_dtypes(include=["int64", "float64"]).columns
numeric_cols = [col for col in numeric_cols if col not in ["salary", "salary_normalized"]]  # éviter doublons

# Remplacement des NaN par 0 pour éviter les erreurs PCA
df_pca = df[numeric_cols].fillna(0)

# Application PCA pour réduire à 2 dimensions
pca = PCA(n_components=2)
pca_result = pca.fit_transform(df_pca)

# Ajout des colonnes PCA au dataset
df["PCA1"] = pca_result[:, 0]
df["PCA2"] = pca_result[:, 1]

print("\nVariance expliquée par les 2 composantes principales :", pca.explained_variance_ratio_)

# Optionnel : t-SNE (plus lent mais utile pour visualisation non linéaire)
# tsne = TSNE(n_components=2, random_state=42)
# tsne_result = tsne.fit_transform(df_pca)
# df["tSNE1"], df["tSNE2"] = tsne_result[:, 0], tsne_result[:, 1]

# 4) Agrégation par niveau d'expérience
salary_stats = df.groupby("experience_level").agg(
    avg_salary=("salary", "mean"),
    median_salary=("salary", "median")
).reset_index()

print("\nMoyenne et médiane des salaires par niveau d’expérience :")
print(salary_stats)

# 5) Résumé final
print("\nColonnes disponibles après traitement :")
print(df.columns.tolist())

# Sauvegarde optionnelle des résultats
# df.to_csv("ds_salary_processed.csv", index=False)
```
