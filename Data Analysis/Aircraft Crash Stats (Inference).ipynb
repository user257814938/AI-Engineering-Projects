```python
# 1. Import Libraries
import pandas as pd
import numpy as np
from scipy import stats
import matplotlib.pyplot as plt
import seaborn as sns

# 2. Load Dataset
df = pd.read_csv("Airplane_Crashes_and_Fatalities_upto_2023.csv")

# 3. Clean Data
df.columns = df.columns.str.lower().str.replace(" ", "_")
df["date"] = pd.to_datetime(df["date"], errors="coerce")
df = df.dropna(subset=["date"])
df["year"] = df["date"].dt.year
df["decade"] = (df["year"] // 10) * 10
df["fatalities"] = pd.to_numeric(df["fatalities"], errors="coerce").fillna(0)
df["aboard"] = pd.to_numeric(df["aboard"], errors="coerce").fillna(0)

# 4. Basic Statistics
print("Total crashes:", len(df))
print("Total fatalities:", int(df["fatalities"].sum()))
print("Average fatalities per crash:", round(df["fatalities"].mean(), 2))

# 5. Crashes per Year
crashes_per_year = df.groupby("year").size()
crashes_per_year.plot(title="Crashes per Year")
plt.show()

# 6. Fatalities Distribution
sns.histplot(df["fatalities"], bins=30, kde=True)
plt.title("Fatalities Distribution")
plt.show()

# 7. Hypothesis Testing Between Decades
if "decade" in df.columns:
    decades = df["decade"].value_counts().index.tolist()
    if len(decades) >= 2:
        g1 = df[df["decade"] == decades[0]]["fatalities"]
        g2 = df[df["decade"] == decades[-1]]["fatalities"]
        t, p = stats.ttest_ind(g1, g2, equal_var=False)
        print("T-test between", decades[0], "and", decades[-1])
        print("t-stat:", t, "p-value:", p)
```
