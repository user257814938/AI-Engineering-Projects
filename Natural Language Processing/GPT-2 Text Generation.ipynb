```python
# Installation des dépendances
!pip install transformers matplotlib torch --quiet

# Imports
import warnings
warnings.filterwarnings('ignore')

from transformers import GPT2Tokenizer, GPT2LMHeadModel
import torch
import matplotlib.pyplot as plt
import numpy as np

print('='*50)
print('EXERCICE 1: What Are Large Language Models (LLMs)?')
print('='*50)

# Explication sur les LLMs
explanation_llm = """
Les LLMs (Large Language Models) sont des modeles de reseau de neurones entraines
sur de vastes quantites de texte. Ils peuvent generer du texte, repondre a des questions,
traduire des langues, et bien plus.

Caracteristiques principales:
- Bases sur l'architecture Transformer
- Entraines sur des milliards de parametres
- Utilisent l'auto-attention pour comprendre les relations entre les mots
- Peuvent effectuer diverses taches de traitement du langage naturel
"""

print('Explication des LLMs:')
print(explanation_llm)

# Chargement du modele GPT-2
print('\n' + '='*50)
print('Chargement du modele GPT-2...')
tokenizer = GPT2Tokenizer.from_pretrained('gpt2')
model = GPT2LMHeadModel.from_pretrained('gpt2')
print('Model GPT-2 charge avec succes')
print(f'Nombre de parametres du modele: {sum(p.numel() for p in model.parameters()):,}')

print('\n' + '='*50)
print('EXERCICE 2: Transformer Architecture And Tokenization')
print('='*50)

# Explication de la tokenization
print('\nExercice 2: Tokenization')
print('La tokenization est le processus de conversion du texte en tokens.')
print('Chaque token represente une unite semantique (mot, sous-mot, caractere).')

# Exemple de tokenization
sample_text = "Bonjour, comment ca va aujourd'hui?"
tokens = tokenizer.encode(sample_text)
token_ids = tokenizer.encode(sample_text)
token_strings = tokenizer.convert_ids_to_tokens(token_ids)

print(f'\nTexte original: {sample_text}')
print(f'Tokens: {token_strings}')
print(f'IDs des tokens: {token_ids}')

# Visualization de la tokenization
print('\n' + '='*50)
print('EXERCICE 3: Understanding Token IDs And Special Prefixes')
print('='*50)

# Token IDs et prefixes speciaux
print('\nExercice 3: IDs des tokens et prefixes speciaux')
print('Les tokens speciaux ont des IDs particuliers:')

special_tokens = {
    'BOS (Beginning of Sequence)': tokenizer.bos_token_id,
    'EOS (End of Sequence)': tokenizer.eos_token_id,
    'PAD (Padding)': tokenizer.pad_token_id,
    'UNK (Unknown)': tokenizer.unk_token_id,
}

for token_name, token_id in special_tokens.items():
    print(f'{token_name}: {token_id}')

# Demonstration avec un texte complet
test_sentence = 'Transformers are powerful!'
test_ids = tokenizer.encode(test_sentence)
test_tokens = tokenizer.convert_ids_to_tokens(test_ids)

print(f'\nPhrase: {test_sentence}')
for token, token_id in zip(test_tokens, test_ids):
    print(f'  Token: \"{token}\" -> ID: {token_id}')

print('\n' + '='*50)
print('EXERCICE 4: Pretraining Vs. Fine-Tuning')
print('='*50)

pretraining_explanation = """
Difference entre Pretraining et Fine-Tuning:

PRETRAINING:
- Phase initiale d'entrainement sur de tres grandes quantites de texte
- Le modele apprend les motifs linguistiques generaux
- Utilise des taches auto-supervisees (prediction du mot suivant)
- Prend des semaines/mois sur des GPUs puissantes

FINE-TUNING:
- Adaptation du modele preentraine a une tache specifique
- Utilise des donnees etiquetees de petite taille
- Plus rapide et efficace que pretraining
- Preserve les connaissances generales du pretraining
"""

print(pretraining_explanation)

print('\n' + '='*50)
print('EXERCICE 5: Generate Simple Text')
print('='*50)

# Generation de texte simple
print('\nExercice 5: Generation de texte')
print('Nous allons generer du texte simple avec GPT-2')

# Input prompt
prompt = "The future of artificial intelligence"
print(f'\nPrompt: \"{prompt}\"')
print('\nTexte genere:')

# Encoder le prompt
input_ids = tokenizer.encode(prompt, return_tensors='pt')

# Generer du texte
output = model.generate(
    input_ids,
    max_length=50,
    num_return_sequences=1,
    no_repeat_ngram_size=2,
    temperature=0.7,
    top_p=0.9,
)

# Decoder la sortie
generated_text = tokenizer.decode(output[0], skip_special_tokens=True)
print(generated_text)

# Generation alternative avec un autre prompt
alt_prompt = "Machine learning"
alt_input_ids = tokenizer.encode(alt_prompt, return_tensors='pt')
alt_output = model.generate(
    alt_input_ids,
    max_length=50,
    num_return_sequences=1,
    temperature=0.6,
    top_p=0.95,
)
alt_generated_text = tokenizer.decode(alt_output[0], skip_special_tokens=True)
print(f'\nAlternative prompt: \"{alt_prompt}\"')
print(f'Texte genere: {alt_generated_text}')

print('\n' + '='*50)
print('Tous les exercices ont ete completes avec succes!')
print('='*50)
```
