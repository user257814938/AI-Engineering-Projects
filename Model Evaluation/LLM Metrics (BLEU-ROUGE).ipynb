```python
# Evaluating Large Language Models
# Toutes les 6 tâches dans une seule cellule

import nltk
from nltk.translate.bleu_score import sentence_bleu, SmoothingFunction
from rouge_score import rouge_scorer
import numpy as np

print("=== DAILY CHALLENGE - Evaluating Large Language Models ===")

# TASK 1: Understanding LLM Evaluation
print("\n=== TASK 1: Understanding LLM Evaluation ===")
print("\n1. Why evaluating LLMs is more complex than traditional software:")
print("   - LLMs generate probabilistic outputs with multiple valid responses")
print("   - No single 'correct' answer for creative/open-ended tasks")
print("   - Context-dependent and subjective quality measures")
print("   - Must evaluate fluency, coherence, factuality, bias, safety simultaneously")

print("\n2. Key reasons for evaluating LLM safety:")
print("   - Prevent harmful, toxic, or biased content generation")
print("   - Ensure alignment with human values and ethical standards")
print("   - Detect potential security vulnerabilities (prompt injection, data leakage)")
print("   - Maintain user trust and regulatory compliance")

print("\n3. How adversarial testing contributes to LLM improvement:")
print("   - Identifies edge cases and failure modes")
print("   - Reveals biases and robustness issues")
print("   - Tests model behavior under malicious inputs")
print("   - Guides targeted improvements and safety fine-tuning")

print("\n4. Limitations of automated metrics vs human evaluation:")
print("   - Automated metrics (BLEU/ROUGE): Fast but miss semantic meaning, creativity, context")
print("   - Can't evaluate factual accuracy, common sense, or appropriateness")
print("   - Human evaluation: Captures nuance, context, quality but slow and expensive")
print("   - Best approach: Combine both for comprehensive evaluation")

# TASK 2: Applying BLEU and ROUGE Metrics
print("\n=== TASK 2: BLEU and ROUGE Metrics ===")

# BLEU Score Calculation
reference_bleu = "Despite the increasing reliance on artificial intelligence in various industries, human oversight remains essential to ensure ethical and effective implementation."
generated_bleu = "Although AI is being used more in industries, human supervision is still necessary for ethical and effective application."

ref_tokens_bleu = reference_bleu.lower().split()
gen_tokens_bleu = generated_bleu.lower().split()

smoothing = SmoothingFunction().method1
bleu_score = sentence_bleu([ref_tokens_bleu], gen_tokens_bleu, smoothing_function=smoothing)

print(f"\nBLEU Score Calculation:")
print(f"Reference: {reference_bleu}")
print(f"Generated: {generated_bleu}")
print(f"BLEU Score: {bleu_score:.4f}")

# ROUGE Score Calculation
reference_rouge = "In the face of rapid climate change, global initiatives must focus on reducing carbon emissions and developing sustainable energy sources to mitigate environmental impact."
generated_rouge = "To counteract climate change, worldwide efforts should aim to lower carbon emissions and enhance renewable energy development."

scorer = rouge_scorer.RougeScorer(['rouge1', 'rouge2', 'rougeL'], use_stemmer=True)
rouge_scores = scorer.score(reference_rouge, generated_rouge)

print(f"\nROUGE Score Calculation:")
print(f"Reference: {reference_rouge}")
print(f"Generated: {generated_rouge}")
for metric, scores in rouge_scores.items():
    print(f"{metric.upper()}: Precision={scores.precision:.4f}, Recall={scores.recall:.4f}, F1={scores.fmeasure:.4f}")

print("\nLimitations of BLEU and ROUGE for creative/context-sensitive text:")
print("   - Focus on n-gram overlap, missing semantic similarity")
print("   - Don't capture paraphrasing, synonyms well (though ROUGE uses stemming)")
print("   - Ignore context, coherence, and factual accuracy")
print("   - Poor for creative text where exact wording varies but meaning preserved")
print("   - Can't evaluate originality, style, or appropriateness")

print("\nSuggested improvements/alternatives:")
print("   - BERTScore: Uses contextual embeddings for semantic similarity")
print("   - Human evaluation: For creativity, coherence, appropriateness")
print("   - Task-specific metrics: Perplexity for language modeling, accuracy for QA")
print("   - Multi-dimensional scoring: Separate scores for different quality aspects")
print("   - LLM-as-judge: Use GPT-4 to evaluate other models' outputs")

# TASK 3: Perplexity Analysis
print("\n=== TASK 3: Perplexity Analysis ===")

model_a_prob = 0.8
model_b_prob = 0.4

perplexity_a = 1 / model_a_prob
perplexity_b = 1 / model_b_prob

print(f"\nModel A: P(mitigation) = {model_a_prob} => Perplexity = {perplexity_a:.4f}")
print(f"Model B: P(mitigation) = {model_b_prob} => Perplexity = {perplexity_b:.4f}")
print(f"\nModel A has LOWER perplexity ({perplexity_a:.2f} < {perplexity_b:.2f})")
print("\nExplanation:")
print("   - Lower perplexity = better model (more confident predictions)")
print("   - Model A assigns higher probability to correct word")
print("   - Perplexity measures uncertainty: lower is better")
print("   - Model A is less 'perplexed' by the data")

print("\nPerplexity = 100 implications:")
print("   - Model is equivalent to randomly choosing from 100 words")
print("   - Relatively high perplexity indicates poor performance")
print("   - Model has high uncertainty about predictions")

print("\nWays to improve perplexity of 100:")
print("   - Increase model size (more parameters for better patterns)")
print("   - Train on more diverse, high-quality data")
print("   - Improve preprocessing and tokenization")
print("   - Use better architecture (Transformer > LSTM for long-range)")
print("   - Optimize hyperparameters (learning rate, batch size)")
print("   - Regularization to prevent overfitting")
print("   - Domain-specific fine-tuning")

# TASK 4: Human Evaluation Exercise
print("\n=== TASK 4: Human Evaluation ===")

chatbot_response = "Apologies, but comprehend I do not. Could you rephrase your question?"

print(f"\nChatbot Response: '{chatbot_response}'")
print("\nFluency Rating (Likert 1-5): 2/5")
print("\nJustification:")
print("   - Unnatural word order ('comprehend I do not' - Yoda-like syntax)")
print("   - Grammatically awkward structure")
print("   - Meaning is clear but expression is poor")
print("   - Not typical conversational English")

improved_response = "I apologize, but I don't understand. Could you please rephrase your question?"
print(f"\nImproved Response: '{improved_response}'")
print("\nWhy it's better:")
print("   - Natural word order and grammar")
print("   - Professional yet friendly tone")
print("   - Clear and concise communication")
print("   - Standard conversational English")
print("   - Maintains politeness while being fluent")

# TASK 5: Adversarial Testing Exercise
print("\n=== TASK 5: Adversarial Testing ===")

prompt_adversarial = "What is the capitol of France?"
expected = "Paris."

print(f"\nPrompt: '{prompt_adversarial}'")
print(f"Expected: '{expected}'")
print("\nPotential mistake:")
print("   - LLM might respond with 'The capitol of France is Paris'")
print("   - But the correct answer should be: 'The CAPITAL of France is Paris'")
print("   - 'Capitol' (building) vs 'Capital' (city) - spelling error in prompt")
print("   - Robust model should either:")
print("     1. Correct the spelling and answer appropriately")
print("     2. Clarify the distinction between capitol/capital")

print("\nMethod to improve robustness:")
print("   - Spell-checking and error correction in preprocessing")
print("   - Training on noisy/misspelled data")
print("   - Prompt understanding and clarification mechanisms")
print("   - Fine-tuning with adversarial examples")
print("   - Implement confidence thresholds for ambiguous inputs")

print("\n3 Tricky Prompts:")
tricky_prompts = [
    {
        "prompt": "How many 'r's are in the word 'strawberry'?",
        "challenge": "Tests character counting - LLMs often fail at exact character-level tasks",
        "expected": "3 (st-r-awbe-r--r-y)"
    },
    {
        "prompt": "If a doctor gives you 3 pills and tells you to take one every half hour, how long do the pills last?",
        "challenge": "Tests logical reasoning - correct answer is 1 hour, not 1.5 hours",
        "expected": "1 hour (0min: pill 1, 30min: pill 2, 60min: pill 3)"
    },
    {
        "prompt": "Complete this sequence: Woman, Man, Camera, Person, __",
        "challenge": "Tests bias/factual knowledge - reference to Trump's cognitive test, may reveal political bias",
        "expected": "TV (from the cognitive test meme)"
    }
]

for i, test in enumerate(tricky_prompts, 1):
    print(f"\nTricky Prompt {i}:")
    print(f"   Prompt: {test['prompt']}")
    print(f"   Challenge: {test['challenge']}")
    print(f"   Expected: {test['expected']}")

# TASK 6: Comparative Analysis of Evaluation Methods
print("\n=== TASK 6: Comparative Analysis ===")

task_chosen = "Machine Translation"

print(f"\nChosen NLP Task: {task_chosen}")
print("\nComparison of 3 Evaluation Metrics:")

metrics_comparison = [
    {
        "metric": "BLEU",
        "strengths": [
            "Fast, automatic, reproducible",
            "Good for comparing similar translations",
            "Correlates reasonably with human judgment for MT",
            "Industry standard for MT benchmarking"
        ],
        "weaknesses": [
            "Focuses on precision, not recall",
            "Requires exact n-gram matches",
            "Poor for single sentence evaluation",
            "Doesn't capture semantic equivalence well"
        ]
    },
    {
        "metric": "BERTScore",
        "strengths": [
            "Captures semantic similarity via embeddings",
            "Better handles paraphrases and synonyms",
            "More robust to word order changes",
            "Correlates better with human judgment than BLEU"
        ],
        "weaknesses": [
            "Computationally expensive (requires model inference)",
            "Slower than BLEU/ROUGE",
            "Requires specific BERT model choice",
            "May not capture domain-specific terminology well"
        ]
    },
    {
        "metric": "Human Evaluation",
        "strengths": [
            "Gold standard for quality assessment",
            "Captures fluency, adequacy, cultural nuances",
            "Can evaluate context appropriateness",
            "Identifies subtle errors automated metrics miss"
        ],
        "weaknesses": [
            "Expensive and time-consuming",
            "Subjective and inconsistent between evaluators",
            "Not scalable for large datasets",
            "Requires expertise in both languages"
        ]
    }
]

for metric_info in metrics_comparison:
    print(f"\n{metric_info['metric']}:")
    print("   Strengths:")
    for strength in metric_info['strengths']:
        print(f"      - {strength}")
    print("   Weaknesses:")
    for weakness in metric_info['weaknesses']:
        print(f"      - {weakness}")

print("\nMost Appropriate Metric for Machine Translation: BLEU + BERTScore")
print("\nReason:")
print("   - BLEU: Industry standard, fast, good for initial benchmarking")
print("   - BERTScore: Complements BLEU by capturing semantic similarity")
print("   - Together they provide both n-gram precision and semantic coverage")
print("   - Use Human Evaluation for final quality assurance on sample set")
print("   - BLEU for quick comparisons, BERTScore for semantic accuracy")
print("   - Avoid relying solely on any single metric")
print("   - MT specifically benefits from BLEU's precision focus")
print("   - BERTScore handles paraphrasing better than BLEU alone")

print("\n=== DAILY CHALLENGE COMPLETED - 6 TASKS DONE ===")
```
