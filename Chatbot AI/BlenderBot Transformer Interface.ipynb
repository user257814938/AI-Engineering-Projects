# Build a Chatbot with Transformers and Gradio

Building a simple conversational AI using **Hugging Face's Transformers** and **Gradio**. The goal is to load a pre-trained conversational model, maintain the dialogue history, and wrap the interaction in a user-friendly chat interface.

We'll cover the following steps:

1. **Install dependencies** – install the transformers and gradio libraries (this step is for your local environment or Colab; skip if already installed).
2. **Load the BlenderBot model** – use the `facebook/blenderbot-400M-distill` model via the `pipeline` API.
3. **Maintain conversation history** – use the `Conversation` class to store past user inputs and generated responses.
4. **Create a Gradio interface** – define a function that interacts with the model and returns responses, and wrap it using `gr.ChatInterface`.
5. **Launch the app** – run the interface locally to chat with your bot.

> **Note:** The transformers and gradio packages may not be available in this sandbox environment, so this notebook demonstrates the code structure. You should run it in Google Colab or locally with the required packages installed.

```python
from transformers import pipeline, Conversation

# Load the BlenderBot model for conversational AI
chatbot = pipeline('conversational', model='facebook/blenderbot-400M-distill')

# Initialize a Conversation with an opening message
conversation = Conversation("Hello! How can I help you today?")

# Generate an initial response
conversation = chatbot(conversation)
print(conversation)
```

```python
# Function to add a new user message and get the bot's response

def chat_step(user_input: str, conversation: Conversation):
    # Append user input and generate response
    conversation.add_user_input(user_input)
    conversation = chatbot(conversation)
    return conversation

# Example usage:
conversation = chat_step("What can you do?", conversation)
print(conversation)
```

```python
import gradio as gr

# Define a function for Gradio that interacts with the model

def mini_chatbot(message, history):
    # Recreate a conversation with previous turns
    conv = Conversation(
        text=message,
        past_user_inputs=[item[0] for item in history],
        generated_responses=[item[1] for item in history],
    )
    conv = chatbot(conv)
    # Return the last generated response
    response = conv.generated_responses[-1]
    return response

# Create a ChatInterface for the mini_chatbot
chat_iface = gr.ChatInterface(
    fn=mini_chatbot,
    title="Transformers Chatbot",
    description="A simple BlenderBot chatbot using Hugging Face Transformers and Gradio.",
)

# Launch the interface (uncomment to run locally)
# chat_iface.launch()
```
