# Personalized AI Greeter

we build a mini application that greets users by their name and adapts the greeting based on the time of day. The objective is to practice wrapping a simple Python function with a user interface using either **Gradio** or **Streamlit**. The greeter should say “Good morning”, “Good afternoon” or “Good evening” depending on the current hour and append the person’s name. 

We'll implement:

- A function `greet(name: str) -> str` that inspects the current time and generates a personalized message.
- A Gradio interface via `gr.Interface` that calls this function.  
- (Optional) A Streamlit version showing how you could build the same experience with `st.chat_input` and `st.chat_message`.

You can run this notebook locally or in Google Colab. To test the interface locally, install the necessary libraries using `pip install gradio streamlit`.


```python
from datetime import datetime

def greet(name: str) -> str:
    # Determine the current hour
    current_hour = datetime.now().hour
    if 5 <= current_hour < 12:
        greeting = 'Good morning'
    elif 12 <= current_hour < 17:
        greeting = 'Good afternoon'
    elif 17 <= current_hour < 21:
        greeting = 'Good evening'
    else:
        greeting = 'Hello'
    
    # Clean up the name and build the greeting
    name = name.strip() if name else ''
    return f"{greeting}{', ' + name if name else ''}!"

# Test the greet function directly
print(greet('Alice'))
```

```python
# Gradio Interface for the AI Greeter
# To launch this interface locally, uncomment the last line and run this cell.
import gradio as gr

def main():
    iface = gr.Interface(
        fn=greet,
        inputs=gr.Textbox(lines=1, placeholder='Enter your name...'),
        outputs=gr.Textbox(),
        title='AI Greeter',
        description='Enter your name and receive a personalized greeting based on the current time of day.',
        examples=[["Brian"], ["Ada"]],
    )
    iface.launch()

# main()  # Uncomment to run locally
```

```python
# Optional Streamlit implementation of the AI Greeter
# Save this code to a Python file (e.g., greeter_app.py) and run `streamlit run greeter_app.py` to test.
import streamlit as st
from datetime import datetime

def greet(name: str) -> str:
    current_hour = datetime.now().hour
    if 5 <= current_hour < 12:
        greeting = 'Good morning'
    elif 12 <= current_hour < 17:
        greeting = 'Good afternoon'
    elif 17 <= current_hour < 21:
        greeting = 'Good evening'
    else:
        greeting = 'Hello'
    name = name.strip() if name else ''
    return f"{greeting}{', ' + name if name else ''}!"

st.title('AI Greeter')
st.write('Enter your name to receive a personalized greeting.')

if 'chat_history' not in st.session_state:
    st.session_state['chat_history'] = []

# Display previous interactions
for msg in st.session_state['chat_history']:
    st.write(msg)

name_input = st.text_input('Your name')

if st.button('Greet'):
    response = greet(name_input)
    st.session_state['chat_history'].append(f"You: {name_input}")
    st.session_state['chat_history'].append(f"Bot: {response}")
    st.write(response)

if st.button('Clear Chat'):
    st.session_state['chat_history'] = []
```
